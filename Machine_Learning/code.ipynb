```python
# Leverage neural network to predict the results without using PCA

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam, SGD
from keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('preprocess.csv')
df = df.iloc[70000:]

# Define target variable and features
target = '紫外线（指数）'
labels = df[target]
data = df.drop(columns=[target])

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(data, labels, test_size=0.3, random_state=42)

# Standardize features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Define a function to compile, train, and evaluate the model
def compile_train_evaluate(model, X_train, y_train, X_test, y_test, batch_size=32, learning_rate=0.001, early_stop=True, optimizer='adam', loss='mean_squared_error'):
    if optimizer == 'adam':
        opt = Adam(learning_rate=learning_rate)
    elif optimizer == 'sgd':
        opt = SGD(learning_rate=learning_rate)
    
    model.compile(optimizer=opt, loss=loss)
    callbacks = [EarlyStopping(monitor='val_loss', mode='min', patience=10, verbose=1, restore_best_weights=True)] if early_stop else []
    history = model.fit(X_train, y_train, epochs=500, batch_size=batch_size, validation_split=0.2, callbacks=callbacks, verbose=1)
    mse = model.evaluate(X_test, y_test, verbose=0)
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    epochs_run = len(history.history['loss'])
    return mse, r2, epochs_run, history

# Results storage
results = []

# Baseline Model
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Baseline", mse, r2, epochs_run])

# Ablation 1: Remove one hidden layer
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Remove one hidden layer", mse, r2, epochs_run])

# Ablation 2: Decrease number of neurons
model = Sequential()
model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Decrease number of neurons", mse, r2, epochs_run])

# Ablation 3: Change activation functions
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='tanh'))
model.add(Dense(64, activation='tanh'))
model.add(Dense(32, activation='tanh'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Change activation functions", mse, r2, epochs_run])

# Ablation 4: Increase learning rate
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, learning_rate=0.01)
results.append(["Increase learning rate", mse, r2, epochs_run])

# Ablation 5: Decrease learning rate
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, learning_rate=0.0001)
results.append(["Decrease learning rate", mse, r2, epochs_run])

# Ablation 6: Remove early stopping
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, early_stop=False)
results.append(["Remove early stopping", mse, r2, epochs_run])

# Ablation 7: Increase batch size
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, batch_size=64)
results.append(["Increase batch size", mse, r2, epochs_run])

# Ablation 8: Decrease batch size
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, batch_size=16)
results.append(["Decrease batch size", mse, r2, epochs_run])


# Convert results to DataFrame and display
results_df = pd.DataFrame(results, columns=["Experiment", "Test MSE", "R^2 Score", "Epochs Run"])
results_df = results_df.round(2)
print(results_df)

# Save results to CSV
results_df.to_csv('ablation_study_results.csv', index=False)

# Plot True vs Predicted values for Baseline Model
plt.figure(figsize=(10, 6))
y_test, y_pred = y_test, model.predict(X_test)
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel('True Values')
plt.ylabel('Predictions')
plt.title('True Values vs Predictions (Baseline)')
plt.show()

# Plot training & validation loss values for Baseline Model
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss (Baseline)')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper right')
plt.show()
```

```python
# Leverage neural network to predict the results using PCA

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from sklearn.decomposition import PCA
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam, SGD
from keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('preprocess.csv')
df = df.iloc[70000:]

# Define target variable and features
target = '紫外线（指数）'
labels = df[target]
data = df.drop(columns=[target])

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(data, labels, test_size=0.3, random_state=42)

# Standardize features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Apply PCA to the dataset
pca = PCA(n_components=0.95)  
X_train = pca.fit_transform(X_train)
X_test = pca.transform(X_test)

# Define a function to compile, train, and evaluate the model
def compile_train_evaluate(model, X_train, y_train, X_test, y_test, batch_size=32, learning_rate=0.001, early_stop=True, optimizer='adam', loss='mean_squared_error'):
    if optimizer == 'adam':
        opt = Adam(learning_rate=learning_rate)
    elif optimizer == 'sgd':
        opt = SGD(learning_rate=learning_rate)
    
    model.compile(optimizer=opt, loss=loss)
    callbacks = [EarlyStopping(monitor='val_loss', mode='min', patience=10, verbose=1, restore_best_weights=True)] if early_stop else []
    history = model.fit(X_train, y_train, epochs=500, batch_size=batch_size, validation_split=0.2, callbacks=callbacks, verbose=1)
    mse = model.evaluate(X_test, y_test, verbose=0)
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    epochs_run = len(history.history['loss'])
    return mse, r2, epochs_run, history

# Results storage
results = []

# Baseline Model
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Baseline", mse, r2, epochs_run])

# Ablation 1: Remove one hidden layer
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Remove one hidden layer", mse, r2, epochs_run])

# Ablation 2: Decrease number of neurons
model = Sequential()
model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Decrease number of neurons", mse, r2, epochs_run])

# Ablation 3: Change activation functions
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='tanh'))
model.add(Dense(64, activation='tanh'))
model.add(Dense(32, activation='tanh'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test)
results.append(["Change activation functions", mse, r2, epochs_run])

# Ablation 4: Increase learning rate
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, learning_rate=0.01)
results.append(["Increase learning rate", mse, r2, epochs_run])

# Ablation 5: Decrease learning rate
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, learning_rate=0.0001)
results.append(["Decrease learning rate", mse, r2, epochs_run])

# Ablation 6: Remove early stopping
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, early_stop=False)
results.append(["Remove early stopping", mse, r2, epochs_run])

# Ablation 7: Increase batch size
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, batch_size=64)
results.append(["Increase batch size", mse, r2, epochs_run])

# Ablation 8: Decrease batch size
model = Sequential()
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))
mse, r2, epochs_run, history = compile_train_evaluate(model, X_train, y_train, X_test, y_test, batch_size=16)
results.append(["Decrease batch size", mse, r2, epochs_run])


# Convert results to DataFrame and display
results_df = pd.DataFrame(results, columns=["Experiment", "Test MSE", "R^2 Score", "Epochs Run"])
results_df = results_df.round(2)
print(results_df)

# Save results to CSV
results_df.to_csv('ablation_study_results_PCA.csv', index=False)

# Plot True vs Predicted values for Baseline Model
plt.figure(figsize=(10, 6))
y_test, y_pred = y_test, model.predict(X_test)
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel('True Values')
plt.ylabel('Predictions')
plt.title('True Values vs Predictions (Baseline)')
plt.show()

# Plot training & validation loss values for Baseline Model
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss (Baseline)')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper right')
plt.show()
```
